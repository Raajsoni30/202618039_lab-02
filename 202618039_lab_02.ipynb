```python
# Create two matrices
matrix_A = np.array([
    [1, 2],
    [3, 4]
])

matrix_B = np.array([
    [5, 6],
    [7, 8]
])

# Display the matrices
print("Matrix A:")
print(matrix_A)

print("\nMatrix B:")
print(matrix_B)

# Matrix addition
addition = matrix_A + matrix_B

print("\nMatrix Addition:")
print(addition)

# Element-wise multiplication
element_wise_multiplication = matrix_A * matrix_B

print("\nElement-wise Multiplication:")
print(element_wise_multiplication)

# Matrix multiplication
matrix_multiplication = matrix_A @ matrix_B

print("\nMatrix Multiplication:")
print(matrix_multiplication)
```

Matrix addition adds the corresponding elements of the two matrices. Element-wise multiplication multiplies corresponding elements of the two matrices. Matrix multiplication using the @ operator follows the rules of matrix multiplication, where rows of the first matrix are multiplied with columns of the second matrix.

```python
# Create a square matrix
square_matrix = np.array([
    [1, 2],
    [3, 4]
])

print("Square Matrix:")
print(square_matrix)

# Find the transpose
transpose_matrix = square_matrix.T

print("\nTranspose of the Matrix:")
print(transpose_matrix)

# Find the determinant
determinant = np.linalg.det(square_matrix)

print("\nDeterminant of the Matrix:")
print(determinant)
```

The transpose of a matrix is obtained by interchanging its rows and columns. The determinant of the given square matrix is calculated using np.linalg.det(). For the given matrix, the determinant is -2.

```python
# Create a square matrix
A = np.array([
    [1, 2],
    [3, 4]
])

print("Matrix A:")
print(A)

# Find the inverse of the matrix
A_inverse = np.linalg.inv(A)

print("\nInverse of Matrix A:")
print(A_inverse)

# Verify A @ A_inverse
result = A @ A_inverse

print("\nA @ A_inverse:")
print(result)

# Check whether the result is approximately equal to the identity matrix
identity_matrix = np.eye(A.shape[0])

verification = np.allclose(result, identity_matrix)

print("\nIs A @ A_inverse approximately equal to the identity matrix?")
print(verification)
```


The given matrix is invertible because its determinant is non-zero. Its inverse was calculated using np.linalg.inv(). The multiplication of the matrix by its inverse gives a result approximately equal to the identity matrix. The np.allclose() function returns True, confirming that the inverse was verified successfully.

```python
# Vectorized operations on matrices

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Vectorized addition
addition = A + B

# Vectorized element-wise multiplication
multiplication = A * B

# Vectorized matrix multiplication
matrix_multiplication = A @ B

print("Vectorized Addition:")
print(addition)

print("\nVectorized Element-wise Multiplication:")
print(multiplication)

print("\nVectorized Matrix Multiplication:")
print(matrix_multiplication)
```

NumPy vectorized operations perform calculations on entire arrays or matrices without using explicit Python loops. This makes the code shorter, simpler, and efficient for numerical computations.

The table shows both the count and percentage of missing values for each column.

```python
# Step 1: Show missing counts before filling
print("Missing values before filling:")
print(df['Age'].isnull().sum())

# Step 2: Fill missing Age values with mean Age
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)

# Step 3: Show missing counts after filling
print("\nMissing values after filling:")
print(df['Age'].isnull().sum())
```

Before filling: Age column had missing values (around 177 in Titanic dataset).

After filling: Missing values = 0.

This is a common imputation technique to handle missing data, ensuring the dataset is complete for analysis

```python
print("Missing Age values before:", df['Age'].isnull().sum())

# 1. Fill with Mean
df_mean = df.copy()
df_mean['Age'] = df_mean['Age'].fillna(df_mean['Age'].mean())

# 2. Fill with Median
df_median = df.copy()
df_median['Age'] = df_median['Age'].fillna(df_median['Age'].median())

# 3. Fill with Mode (most frequent value)
df_mode = df.copy()
df_mode['Age'] = df_mode['Age'].fillna(df_mode['Age'].mode()[0])

# 4. Fill with Random Value (sampled from existing Age values)
df_random = df.copy()
random_values = np.random.choice(df_random['Age'].dropna(), size=df_random['Age'].isnull().sum())
df_random.loc[df_random['Age'].isnull(), 'Age'] = random_values

# Show missing counts after each method
print("Missing Age after Mean:", df_mean['Age'].isnull().sum())
print("Missing Age after Median:", df_median['Age'].isnull().sum())
print("Missing Age after Mode:", df_mode['Age'].isnull().sum())
print("Missing Age after Random:", df_random['Age'].isnull().sum())
```

Mean imputation smooths values but can distort distribution if outliers exist.

Median imputation is more robust when data is skewed.

Mode imputation works well for categorical or discrete values.

Random imputation keeps variability but introduces randomness.

```python
# Calculate Q1 and Q3
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate the 1.5 × IQR bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
fare_outliers = df[
    (df['Fare'] < lower_bound) |
    (df['Fare'] > upper_bound)
]

# Count outliers
outlier_count = fare_outliers.shape[0]

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", outlier_count)
```

```python
# Create FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Create IsAlone
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

# Display the new columns
print(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head(10))
```

The FamilySize feature represents the total number of family members travelling with each passenger, including the passenger themselves. The IsAlone feature assigns 1 to passengers travelling alone and 0 to passengers travelling with one or more family members. This creates two useful features for further analysis of passenger survival.

```python
# Create pivot table
survival_pivot = pd.pivot_table(
    df,
    index='Sex',
    columns='Pclass',
    values='Survived',
    aggfunc='mean'
)

print("Survival Rate Pivot Table:")
print(survival_pivot)

# Find the highest survival rate
highest_group = survival_pivot.stack().idxmax()
highest_rate = survival_pivot.stack().max() * 100

# Find the lowest survival rate
lowest_group = survival_pivot.stack().idxmin()
lowest_rate = survival_pivot.stack().min() * 100

print("\nHighest Survival Rate:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate, "%")

print("\nLowest Survival Rate:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate, "%")
```

The pivot table shows the mean survival rate for each combination of passenger sex and passenger class. The highest and lowest survival-rate groups are identified from the pivot table using idxmax() and idxmin(). The survival rate is multiplied by 100 to express it as a percentage.

```python
import seaborn  as sns
# Select relevant numerical columns
numerical_data = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'FamilySize', 'IsAlone']]

# Calculate correlation matrix
correlation_matrix = numerical_data.corr()

# Plot correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Titanic Numerical Features')
plt.show()

# Find strongest positive and negative relationships
corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack()

strongest_positive = corr_pairs.idxmax()
positive_value = corr_pairs.max()

strongest_negative = corr_pairs.idxmin()
negative_value = corr_pairs.min()

print("Strongest Positive Relationship:")
print(strongest_positive, ":", positive_value)

print("\nStrongest Negative Relationship:")
print(strongest_negative, ":", negative_value)
```
